# How are poles and zeroes of $H(z)$ influencing $|(H(e^{j\omega})|$
## Reminders
- For IIR Filters
  $y[n] = \sum a_{k}y[n-k]0\sum b_{k}x[n-k]$ -> Z-Transform $h(z)=\frac{y(z)}{x(z)}$ fyll ut från föreläsning

- $|H(e^{i\omega})|=H(z)|\to z=e^{j\omega}$

Lets assume that our $H(z)$ is written in the following form
$$
H(z) = G_1 \frac{(z-z_{1})(z-z_{2})\dots(z-z_{M})}{(z-p_{1})(z-p_{2})\dots(z-p_{N})}
\to 
|H(e^{j\omega})| =G_1 \frac{(e^{j\omega}-z_{1})(e^{j\omega}-z_{2})\dots(e^{j\omega a}-z_{M})}{(e^{j\omega}-p_{1})(e^{j\omega}-p_{2})\dots(e^{j\omega}-p_{N})}
$$
For simplicity, first a single zero or pole that we call Type equation here. $v_{k}$(since it can be either $z_{k}$ and $p_{k}$) and $|e^{j\omega}-v_{k}|$ is of interest.

A zero close to the unit circle will attenuate $|H(e^{j \omega})|$ near it location (at angular frequencies near its phase). The close the stronger the attentuation (at unit circle: $|H(e^{j \omega})| = 0$)
At pole close to the unit circle will amplify $H(e^{e\omega})$ near its location. The closer the stronger the amplification.

## Example
Draw te pole-zero locaftions for $H(z)$ and try to figure out where 
$H(e^{j\omega})$ is lange and small when
$$
H(z) = 1-\frac{z^{-z}}{1-z^{-1}+\frac{1}{2}z^{-2}}
$$
First break out the highest power of $z^{-1}$ in numerator and denominator

$$
H(z) = \frac{z^{-1}}{z^{-2}}z\frac{z-1}{z²-z+\frac{1}{2}}=z \to z\frac{z-1}{z²-z+\frac{1}{2}}
$$
The first factor contributes with a zero at $z_{1} = 0$ and the numerator of the fraction another on at $z_{2} = 1$

Finding the roots of the second degree polynomial in the denominator gives to two conjugate poles at $p_{1}=0.5(1+j)$ and $p_2 = 0.5(1-j)$

## Pole-Zero diagram
![[Pasted image 20260216105214.png]]

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

# Transfer Function: H(z) = z(z-1) / (z^2 - z + 0.5)
# Numerator: z^2 - z + 0
b = [1, -1, 0]
# Denominator: z^2 - z + 0.5
a = [1, -1, 0.5]

# Calculate roots
zeros = np.roots(b)
poles = np.roots(a)

# Setup plot
fig, ax = plt.subplots(figsize=(6, 6))

# Draw unit circle
unit_circle = Circle((0, 0), 1, color='black', fill=False, linestyle='--', label='Unit Circle')
ax.add_patch(unit_circle)

# Plot poles and zeros
ax.plot(np.real(zeros), np.imag(zeros), 'go', fillstyle='none', markersize=10, markeredgewidth=2, label='Zeros')
ax.plot(np.real(poles), np.imag(poles), 'rx', markersize=10, markeredgewidth=2, label='Poles')

# Styling
ax.axhline(0, color='gray', lw=1)
ax.axvline(0, color='gray', lw=1)
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.grid(True, linestyle=':', alpha=0.6)
ax.set_xlabel('Real Part')
ax.set_ylabel('Imaginary Part')
ax.set_title(r'Pole-Zero Diagram: $H(z) = \frac{z(z-1)}{z^2 - z + 0.5}$')
ax.legend()

plt.show()
```
