```python
import matplotlib.pyplot as plt
import numpy as np

# Define QPSK symbols (In-phase and Quadrature components)
# Bits: 00, 01, 11, 10
bits = ['00', '01', '11', '10']
points = np.array([1+1j, -1+1j, -1-1j, 1-1j]) / np.sqrt(2)

plt.figure(figsize=(6, 6))
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Plot constellation points
plt.scatter(points.real, points.imag, color='red', s=100)

# Label the points with their bit mapping
for i, bit in enumerate(bits):
    plt.text(points[i].real + 0.05, points[i].imag + 0.05, bit, fontsize=12)

plt.title('QPSK Constellation Diagram')
plt.xlabel('In-phase (I)')
plt.ylabel('Quadrature (Q)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.axis([-1.5, 1.5, -1.5, 1.5])
plt.show()
```
